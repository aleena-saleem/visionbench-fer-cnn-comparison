# VisionBench: CNN Benchmark on FER2013

-This notebook compares the performance of multiple CNN architectures trained on the FER2013 dataset under identical experimental settings.

## Models

- ResNet50
- DenseNet121
- EfficientNet-B0
- ConvNeXt-Tiny

## Evaluation Metrics

- Accuracy
- Precision
- Recall
- F1-score
- Trainable Parameters
- Total Parameters
- Training Strategy (Feature Extraction)

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import pandas as pd

benchmark = pd.DataFrame({

    "Model": [
        "ResNet50",
        "DenseNet121",
        "EfiicientNet-B0",
        "ConvNeXt"
    ],

    "Transfer Learning": [
        "Feature Extraction",
        "Feature Extraction",
        "Feature Extraction",
        "Feature Extraction"
    ],

    "Train Accuracy": [
        44.4,
        42.4,
        42.1,
        51.4
    ],

    "Validation Accuracy": [
        42.6,
        41.6,
        43.4,
        51.3
    ],

    "Test Accuracy": [
        42.3,
        41.5,
        42.3,
        51.3
    ],

    "Precision": [
        40.9,
        38.1,
        40.0,
        49.3
    ],

    "Recall": [
        42.3,
        41.5,
        42.3,
        51.3
    ],

    "F1 Score": [
        39.2,
        38.0,
        39.5,
        49.4
    ],

    "Trainable Params": [
        "14,343",
        "7,175",
        "8,967",
        "5,383"
    ],

    "Total Params": [
        "23.5M",
        "7.98M",
        "4.0M",
        "27.8M"   ]
})

benchmark.to_csv("benchmark_results.csv", index=False)

benchmark
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

labels = (
    benchmark["Model"] +
    "\n(" +
    benchmark["Transfer Learning"] +
    ")"
)

plt.bar(
    labels,
    benchmark["Test Accuracy"]
)

plt.title("VisionBench: CNN Model Comparison on FER2013")
plt.xlabel("Model")
plt.ylabel("Test Accuracy (%)")

for i, value in enumerate(benchmark["Test Accuracy"]):
    plt.text(
        i,
        value + 0.2,
        f"{value:.1f}%",
        ha="center",
        fontsize=10
    )

plt.xticks(rotation=10)

plt.tight_layout()

plt.savefig(
    "benchmark_test_accuracy.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Observations

### ResNet50 (Feature Extraction)

- Highest validation accuracy among the evaluated models.
- Very few trainable parameters.
- Stable training and validation curves.

---

### DenseNet121 (Feature Extraction)

- Achieved performance comparable to ResNet50.
- Required fewer total parameters.
- Lightweight architecture with competitive results.
---


### ConvNeXt (Feature Extraction)


*   Best overall performer — highest train, validation, and test accuracy of all
architectures benchmarked.

*   Smallest trainable parameter count (5,383) despite having the largest total backbone size (~27.8M).

*   Training and validation curves track closely with no signs of overfitting; both curves were still improving at epoch 10, suggesting further gains are possible with more epochs.
